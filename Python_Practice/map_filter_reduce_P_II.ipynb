```python
#Q1.Use map() to create a new list containing the square of each number.
num = [1,2,3,4,5,6]
sq = list(map(lambda n:n**2,num))
sq
```

```python
#Q2. Convert a list of numeric strings into integers using map().
num = ['1','2','3','4','5','6']
nm = list(map(int,num))
nm
```

```python
#Q3. Add corresponding elements of two lists using map().
a = [10,20,30]
b = [5,15,25]
add = list(map(lambda x,y:x+y,a,b))
add
```

```python
#Q4. Use filter() to select only even numbers from a list.
num = [25,64,89,78,22,24,36,55,51,91,28]
even = list(filter(lambda n:n%2==0,num))
even
```

```python
#Q5. Remove empty strings from a list using filter().
a = ['apple','air','','soil','','tree','']
new = list(filter(lambda x:x!='',a))
new
```

```python
#Q6. Filter names whose length is greater than 5 characters.
a = ['Ai','Python','string','ice','bottle','walk']
nm = list(filter(lambda x:len(x)>5,a))
nm
```

```python
#Q7. Use reduce() to find the sum of all numbers in a list.
num = [10,20,30,40,50,60]
from functools import reduce
sum = reduce(lambda x,y:x+y,num)
sum
```

```python
#Q8. use reduce() to multiply all numbers in a list.
num = [10,20,30,40,50,60]
mul = reduce(lambda x,y:x*y,num)
mul
```

```python
#Q9. Find the largest number from a list using reduce().
num = [26,87,96,21,102,136,45]
lr = reduce(lambda x,y:x if x>y else y,num)
lr
```

```python
#Q10. Select even numbers from a list and then square them.
num = [25,64,89,78,22,24,36,55,51,91,28,2]
even = list(filter(lambda n:n%2==0,num))
sq = list(map(lambda n:n**2,even))
print(even)
print(sq)
```

```python
#Q11. Apply a normal function to every element of a list using map().
def add(num):
  return num + 10
numbers = [10,20,30,40,50]
new = list(map(add,numbers))
new
```

```python
# convert celcius to fr
def cl(temp):
  return temp*9/5 +32
temps = [22,36,45,78]
fr = list(map(cl,temps))
fr
```

```python
#Q12. Use reduce() with an initial value to add numbers.
num = [45,36,87,22]
add = reduce(lambda a,b:a+b,num)
add
```
