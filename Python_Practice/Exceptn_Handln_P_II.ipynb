```python
#Q1. Write a Python program to divide two numbers and handle division by zero.
a = 10
b = 0
try:
  div = a/b
  print('Result:',div)
except ZeroDivisionError:
  print('Error: Denominator cannot be zero')
```

```python
#Q2. Write a program that accepts a number from the user and handles invalid input.
try:
  num = int(input('Enter a number:'))
  print('The number entered is:',num)
except ValueError:
  print('Please enter a valid number')
```

```python
#Q3. Write a program that handles ValueError and ZeroDivisionError separately.
a = 10
try:
  b = int(input('Enter a denominator:'))
  print('Result:',a/b)
except ValueError:
  print('Please enter a valid number')
except ZeroDivisionError:
  print('Error: Denominator cannot be zero')
```

```python
#Q4.Write an example using try, except, else, and finally.
try:
  rl = int(input('enter your roll no:'))
except ValueError:
  print('Please enter correct roll no')
else:
  print('Your roll no is:',rl)
finally:
  print('Thank you')
```

```python
#Q5. Write a Python program to handle list index out of range error.
num = [10,20,30,40,50]
try:
  print(num[10])

except IndexError as e:
  print('Error:',e)
```

```python
#Q6. Write a program to handle missing dictionary key error.
a = {'Name':'Om','Age':23,'Place':'Pune'}
try:
  print(a['city'])
except KeyError :
  print('Error: Key not found')
```

```python
#Q7. Write a program that raises an exception if the number is negative.
try:
  num = int(input('Enter a number:'))
  if num < 0 :
    raise ValueError('Number cannot be negative')
  print('The number is:',num)
except ValueError as e:
  print(e)
```

```python
#Q8. Create a custom exception for age below 18.
try:
  age = int(input('enter your age:'))
  if age < 18:
    raise ValueError('Age cannot be less than 18')
  print('You are eligible')
except ValueError as e:
  print(e)
```

```python
#Q9. Write a program to open a file safely using exception handling.
try:
  with open('file.txt','r') as file:
   print(file.read())
except FileNotFoundError:
  print('File not found')
```

```python
#Q10. Write a program showing how finally is used for cleanup.
try:
  a = 45
  b = 0
  div = a/b
  print('result:',div)
except ZeroDivisionError:
  print('Error: Denominator cannot be zero')
finally:
  print('Thank you')
```

```python
#Q11. Create a function safe_divide(a, b) that handles division by zero.
def safe_divide(a,b):
  try:
    return a/b
  except ZeroDivisionError:
   return('Error: Denominator cannot be zero')
print(safe_divide(10,0))
print(safe_divide(10,2))
```
