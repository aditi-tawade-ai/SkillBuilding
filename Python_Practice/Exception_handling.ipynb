Q. Why exception handling is needed?
- While programming we know some function/ statements may return an error during execuition
- Since python is an interprutted language, if an error occur the program stops immediately and the remaining correct lines are not executed
- To avoid this we use Exception handling
- Exception handling catches the error and prints meaning full message insted of stopping the entire program
- This allows the remaining code to continue executing even if error occurs.

```python
#Q1. Handle division by zero so that the program does not crash.
a = 10
b = 0
try:
  result = a/b    # error
  print(result)
except ZeroDivisionError:    # handels error
  print("Cannot divide by zero")
```

```python
#Q2. Use try-except to handle invalid numeric input.
try:
  num = int(input('Enter a number:'))
  print('number entered is:',num)
except ValueError:
  print('Please enter a valid number')
```

```python
#Q3. Handle both invalid input and division by zero in one program.
try:
  a = int(input('Enter a number:'))
  b = int(input('Enter a number:'))
  div = a/b
  print('Result:',div)
except ValueError:
  print('Please enter a valid number')
except ZeroDivisionError:
  print("Cannot divide by zero")
```

```python
#Q4. Catch an exception and display the actual error message.
try:
  num = [10,20,30,40,50]
  print(num(5))
except Exception as e:
  print('Error occured:',e)
```

```python
#Q5. Use else to run code only when no exception occurs.
try:
  num = int(input('Enter a number:'))
except Exception as e:
  print('Error:',e)
else:
  print('Number you entered is:',num)
```

```python
#Q6. Use finally to execute code whether an exception occurs or not.
code = 123
try:
  cd = int(input('Enter the code:'))
  if cd == code:
   print('Welcome')
  else:
    print('Invalid code')
except ValueError:
  print('Please enter a valid number')
finally:
  print('Thank you')
```

```python
#Q8. Create a custom exception for insufficient balance.
class InsufficientBalanceError(Exception):
  pass
amt = 1000
wd = 1500
try:
  if amt < wd :
    raise InsufficientBalanceError('Insufficient balance available')
    amt = amt - wd
  print('withdrawal Successful!')
  print('Remaining amount:',amt)
except InsufficientBalanceError as e:
  print(e)
```

```python
#Q9. Safely read a file and handle the situation when the file does not exist.
try:
  with open('file,txt','r') as file:
    content = file.read()
    print(content)
except FileNotFoundError:
  print('This file does not exist')
```

```python
#Q10. Use nested try-except to handle errors at different levels.
try:
  num = int(input('Enter a number'))
  try:
    div = 100/num
    print('Result:',div)

  except ZeroDivisionError as e:
    print('Error occured:',e)
except ValueError as msg:
  print('Error occured:',msg)
```

```python
#Q11. Handle an error when trying to access an invalid list index.
lst = [1,2,3,4,5]
try:
  print(lst[10])
except IndexError as e:
  print("error occured:",e)
```
