```python
# Write a program that handles ValueError and ZeroDivisionError separately.
a = 10
try:
  b = int(input('enter a number:'))
  print('Result:',a/b)
except ValueError:
  print('Error occured:please enter a valid number')
except ZeroDivisionError:
  print('Error occured: Denominator cannot be zero')
```

```python
# Write a program that raises an exception if the number is negative.
num = int(input('Enter a number:'))
try:
  if num < 0:
    raise ValueError('Number cannot be negative')
    print('Number is:',num)
  else:
    print('Number is:',num)
except ValueError as e:
  print(e)
```

```python
# Write a program to open a file safely using exception handling.
try:
  file = open('file.txt','r')
  print(file.read())
  file.close()
except FileNotFoundError:
  print('File not found')
```

```python
# Write a program showing how finally is used for cleanup.
try:
  a = (input('enter a 4 digit pin:'))
  if not a.isdigit():
   raise ValueError('Pin should not be a alphabetic word')
  print('Your pin is:',a)
except ValueError as e:
  print(e)
finally:
  print('Thank you')
```

```python
# Select even numbers from a list and then square them.
num = [25,14,22,36,39,89,85,13]
evn = list(filter(lambda x:x%2==0,num))
sq = list(map(lambda x:x**2,evn))
print(evn)
print(sq)
```

```python
# Use reduce() to find the sum of all numbers in a list.
from functools import reduce
a = [1,2,3,4,5,6,7,8,9]
mul = reduce(lambda x,y:x*y,a)
print(mul)
```

```python
# Find the largest number from a list using reduce().
a = [26,87,96,21,102,136,45]
lr = reduce(lambda x,y: x if x>y else y, a)
lr
```
