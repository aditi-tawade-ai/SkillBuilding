 Q1. what is lambda function?

-  Lambda function is an ananomous function/ nameless function

- It is used in simple operations with only one condition

- it used ternary operator to iterate if condition


```python
# Q1. Add 10 to a given number
add = lambda x:x+10
add(10)
```

```python
# Q2. Multiply 2 nums
mul = lambda a,b:a*b
mul(5,10)
```

```python
# Q3. return sq of a given num
num = lambda x:x**2
num(3)
```

```python
# Q4. check num is even or odd
num = lambda n:'even' if n%2==0 else 'odd'
print(num(45))
print(num(22))
```

```python
# Q5. use lambda with map to sq every num
num = [1,2,3,4,5,6]
sq = list(map(lambda x:x**2,num))
sq
```

```python
# Q6. use lambda with filter to find only even num
num = [1,2,4,8,9,7,6,3,22,20]
evn = list(filter(lambda x:x%2==0,num))
evn
```

```python
#Q7. sort list of tuple on the second value using lambda
detail = [('Banana',2),('Apple',1),('Grape',5),('Mango',4),('Kiwi',3)]
sort = sorted(detail,key=lambda x:x[1])
sort
```

```python
detail = [('Banana',2),('Apple',1),('Grape',5),('Mango',4),('Kiwi',3)]
sort = sorted(detail,key=lambda x:x[1],reverse=True)
sort
```

```python
#Q8. Sort employee records based on salary list od dict
emp = [{'Om':96000},
       {'Sai':25000},
       {'Shree':45000},
       {'Jui':89000},
       {'maya':1500}]

rec = sorted(emp,key=lambda x:list(x.values()))
rec
```

```python
emp = [{'name':'Om','salary':96000},
       {'name':'Sai','salary':25000},
       {'name':'Shree','salary':45000},
       {'name':'Jui','salary':89000},
       {'name':'maya','salary':1500}]
rec = sorted(emp, key=lambda x:x['salary'])
rec
```

```python
#Q9. return max of 2 num
max = lambda a,b:a if a>b else b
max(45,480)
```

```python
#Q11.Use lambda with reduce() to find the product of all numbers in a list.
# reduce function applies lambda fun to all num and combine to five single o/p
# it 1st mul 1 *2 then 2*3,6*4 and so on...
from functools import reduce
num = [1,2,3,4,5]
mul = reduce(lambda x,y:x*y,num)
mul
```
