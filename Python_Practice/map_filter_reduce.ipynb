```python
#Q1. Square Each Number Using map()
num = [1,2,3,4,5]
sq = list(map(lambda x:x*2,num))
sq
```

```python
#Q2. Convert Celsius to Fahrenheit Using map()
temp = [32,45,78,92]
fr = list(map(lambda x:(x*9/5)+32,temp))
fr
```

```python
#Q3. Extract Names from List of Dictionaries
details = [{1:'Aditi'},{2:'Om'},{3:'sai'},{4:'jui'}]
nms = list(map(lambda x:list(x.values())[0],details))
nms
```

```python
#Q4. Write a Python program to filter only positive numbers from a list.
num = [1,-2,8,-65,12,-78,-45,25]
p = list(filter(lambda n:n>0,num))
p
```

```python
#Q5.Filter students whose marks are greater than or equal to 40.
detail =[
    {'name':'Om','mrk':65},
    {'name':'Jay','mrk':26},
    {'name':'shree','mrk':85},
    {'name':'sayee','mrk':31}
]
mks = list(filter(lambda x:x['mrk']>=40,detail))
mks
```

```python
#Q6.  Filter all words that start with the letter P.
name = ['python','pi','apple','sun','pen','water']
pnm = list(filter(lambda x:x[0]=='p',name))
pnm
```

```python
name = ['python','pi','apple','sun','pen','water']
pnm = list(filter(lambda x:x.startswith('p'),name))
pnm
```

```python
#Q7. Write a Python program to find the sum of list elements using reduce().
from functools import reduce
num = [1,2,3,4,5,6]
sum = reduce(lambda a,b:a+b,num)
sum
```

```python
#Q8.  Find the smallest number from a list using reduce().
num = [54,78,32,21,9,45,26]
sm = reduce(lambda a,b:a if a<b else b,num)
sm
```

```python
#Q9.  Find the total number of characters in a list of words using map() and reduce().
word = ['Apple','Water','air','blue','eye']
char = list(map(lambda x:len(x),word))
total = reduce(lambda a,b:a+b,char)
total
```

```python
#Q10. Use map() and reduce() to apply a 10% discount to prices and calculate the final total bill.
price = [100,200,500,600]
discount = list(map(lambda x:x-(x*0.10),price))
bill = reduce(lambda a,b:a+b,discount)
bill
```

```python
#Q11. From a list of numbers, filter even numbers and find their sum using filter() and reduce()
num = [1,2,8,64,71,36,89,25,47,44,28]
even = list(filter(lambda x:x%2==0,num))
sum = reduce(lambda a,b:a+b,even)
print(even)
print(sum)
```

```python
#Q12. Use lambda with sorted() to sort employees based on salary.
emp = [
    {'Om':65000},
    {'Amit':25000},
    {'Ruhi':96000},
    {'sai':12000}
]
sort = sorted(emp,key = lambda x:list(x.values())[0],reverse=True)
sort
```
