# **Lambda function:**
- It is anonymous function/ nameless function
- Only one expression is allowed
- Lambda by default returns the value of expression

**Syntax:**
- lambda parameters,expression

```python
# Q1. Add 2 numbers using lambda
num = lambda a,b:a+b
num(20,30)
```

```python
# Q2. find sq and cube using lambda
sq = lambda n:n**2
cube = lambda n: n**3
print(sq(5))
print(cube(10))
```

```python
# Q3. check +ve,-ve or 0
num = lambda num:'+ve' if num>0 else '-ve' if num < 0 else 'zero'
num(0)
```

```python
# Q4. Filter words starting with specific letter
nm = ['sayee','sneha','vedant','sarthak','aditi','avani','megha']
result = list(filter(lambda nm:nm[0]=='s',nm))
result
```

```python
# Q5.convert list of strings to upper case
name = ['apple','orange','mango','banana']
result = list(map(lambda name:name.upper(),name))
print(result)
```

```python
# Q6. sort students by mks in decresing order
# syntax sorted: sorted(name of list/tuple/etc, key = functn x:expression, reverse=T/F)
mks = 85,63,21,75,98,46,54
result =sorted(mks, key = lambda x:x,reverse=True)
result
```

```python
# here we have a tuple inside list
# to access both elements of tuple we used sclicing x:x[1]
mks = [('Om',98),('jay',45),('jui',85),('sai',32)]
result = sorted(mks, key = lambda x:x[1] ,reverse=True)
result
```

```python
# Q7. sort employees by name length
# here reverse= false is by default so need to mention

name = ['Shreevardhan','Jay','Sayee','Janvi','Aishwarya','Anushka']
final = sorted(name,key=lambda x:len(x))
final
```

```python
# Q8. Find max salary employee
# max function same syntax as sorted

data = [('Shree',85000),('Jay',45000),('Om',1500),('Jui',145000)]
sort = max(data,key = lambda x:x[1])
print(sort)
```

```python
# Q9. remove empty string from list

a = ['Mango','','Banana','city','AI','','Water']
b = list(filter(lambda x:x!='',a))
b
```

```python
# Q10. calculate final price after discount

final = lambda price,discount:price-(price*discount/100)
final(5000,10)
```

```python
# Q11. use lambda with reduce and find max
num = [45,87,96,32,11,9,28]
from functools import reduce
max = reduce(lambda x,y: x if x > y else y,num)
max
```
