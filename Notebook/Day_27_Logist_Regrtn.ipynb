```python
import pandas as pd

path = 'https://raw.githubusercontent.com/hakimyameen/Datasets/refs/heads/main/heart.csv'

h = pd.read_csv(path)
h[:2]
```

```python
h['target'].value_counts()
```

```python
import seaborn as sns

sns.pairplot(data=h[['age','thalach','cp','target']])
```

```python
# missing data
h.isna().sum()
```

```python
# prepare data
X = h[['age','thalach','cp']]

y = h['target']
```

```python
# splitting data
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
# check score
print("train model score:",model.score(X_train,y_train))
print("test model score:",model.score(X_test,y_test))
```
